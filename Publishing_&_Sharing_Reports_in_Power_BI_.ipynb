1. Why is publishing and sharing reports more important than just building dashboards in Power BI Desktop?

-  Building a dashboard in Power BI Desktop is only the development stage. Publishing and sharing makes the report available to the people who actually need to use the information.

Importance:

Allows management and employees to access reports.
Helps in business decision-making.
Enables collaboration between teams.
Allows reports to be accessed through the Power BI Service.
Provides controlled and secure access to data.

In short: Building creates the report, while publishing and sharing makes the report useful to the organization.

2. Define Power BI Service and explain its role compared to Power BI Desktop.

-   Power BI Service is Microsoft's cloud-based platform used to publish, share, manage, and collaborate on Power BI reports and dashboards.

| Power BI Desktop                             | Power BI Service                                       |
| -------------------------------------------- | ------------------------------------------------------ |
| Used mainly to create reports.               | Used mainly to publish and share reports.              |
| Used for data cleaning and modelling.        | Used for collaboration and distribution.               |
| Creates visualizations and DAX calculations. | Manages workspaces, apps, permissions, and dashboards. |
| Installed on a computer.                     | Cloud-based service.                                   |



Reporting lifecycle:

Connect Data → Transform Data → Model Data → Build Report → Publish → Share → Monitor

3. What does publishing a report mean in Power BI?

-  Publishing means uploading a report created in Power BI Desktop to the Power BI Service.

After publishing:

The report becomes available in a Power BI workspace.
The report can be shared with users or distributed through an App.


Medium Level


4. Steps involved in publishing a report from Power BI Desktop to Power BI Service

The correct sequence is:

Step 1: Create the report

Create visuals, charts, tables, and dashboards in Power BI Desktop.

Step 2: Save the report

Save the Power BI file as a .pbix file.

Step 3: Sign in

Sign in to Power BI Desktop using your organizational Microsoft account.

Step 4: Publish

Click:

Home → Publish

Step 5: Select workspace

Choose the appropriate workspace in Power BI Service.

Step 6: Publish

Click Select/Publish to upload the report.

Step 7: Open Power BI Service

Go to Power BI Service and open the selected workspace.

Step 8: Share or distribute

The report can now be shared directly, placed in an App, or used through workspace access depending on organizational requirements.

6. Difference between Direct Report Sharing, Workspace Access, and Power BI Apps

-   | Method                    | Meaning                                                               | Use Case                                                            |
| ------------------------- | --------------------------------------------------------------------- | ------------------------------------------------------------------- |
| **Direct Report Sharing** | Gives specific users access to an individual report.                  | Sharing a sales report with a small team.                           |
| **Workspace Access**      | Gives users access to a workspace according to their assigned role.   | A BI development team working together on reports.                  |
| **Power BI Apps**         | Packages reports and dashboards into a managed application for users. | Distributing official sales reports to the entire sales department. |


1. Direct Report Sharing

A report owner gives selected users permission to view or interact with a report.

Use case: Sharing a financial report with a manager.

2. Workspace Access

Users are given access to a workspace with a specific role such as Viewer, Contributor, Member, or Admin.

Use case: A BI team collaborates on developing and maintaining reports.

3. Power BI Apps

An App provides a controlled package of reports and dashboards for a particular audience.

Use case: Publishing an official Company Sales App for sales employees.

7. Describe the four workspace roles in Power BI

-   Power BI provides four main workspace roles:

1. Admin
Has the highest level of workspace permissions.
Can manage workspace settings.
Can add/remove users.
Can manage access and content.
2. Member
Can publish and manage content.
Can share content and collaborate with other users.
Has more permissions than a Contributor.
3. Contributor
Can create, edit, and publish content.
Can work with reports and semantic models.
Has fewer management permissions than Members/Admins.
4. Viewer
Mainly used to view and interact with content.
Cannot modify reports or workspace content.
Access Control

Organizations assign users the minimum role required for their work.

For example:

BI Manager → Admin
BI Developer → Member/Contributor
Business User → Viewer

This follows the principle of least privilege, reducing unnecessary access.

8. Why are Power BI Apps more secure and scalable than direct sharing?

-   Power BI Apps are useful for enterprise environments because they provide a controlled and organized way to distribute content.

Governance advantages:
Centralized distribution
Reports can be packaged and distributed to a defined audience.
Consistent content
Users receive an official version of the reports instead of multiple individually shared copies.
Easier management
Administrators can manage the content and audience centrally.
Scalability
An App can be distributed to a large number of users without individually sharing every report.
Better governance
Organizations can separate report development from report consumption.
Example:

Instead of individually sharing a sales report with 500 employees, the organization can publish a Sales App for the sales department.

Therefore, Apps provide better centralized distribution, governance, and scalability than repeatedly using direct sharing.

9. Explain Row-Level Security (RLS) in Power BI

-    Row-Level Security (RLS) is a Power BI security feature that restricts the rows of data a user can see.

Different users can use the same report, but each user sees only the data they are authorized to access.

Example:

Suppose a company has sales data for:

India
USA
UK
Australia

An organization can create an RLS rule based on Region.

| User              | Data Visible |
| ----------------- | ------------ |
| India Manager     | India        |
| USA Manager       | USA          |
| UK Manager        | UK           |
| Australia Manager | Australia    |


The same report is used by everyone, but Power BI filters the underlying data according to the user's role.

Basic process:

Create Role → Define Filter → Assign Users → User Opens Report → RLS Filters Data

For example, a role could have a filter:

[Country] = "India"

For larger organizations, dynamic RLS can use the logged-in user's identity and a mapping table to determine which data they should see.

10. Risks of sharing reports without Apps, Workspace Roles, or RLS

-    If an organization shares reports without proper security and governance, several problems can occur.

1. Unauthorized Data Access

Users may see information they are not supposed to access.

Example:
An employee from the India region could potentially see confidential sales information for other countries.

Best practice:
Use RLS to restrict data according to users or regions.

2. Poor Governance and Uncontrolled Content

If reports are individually shared without proper management, an organization may end up with many versions of the same report.

Problems:

Duplicate reports.
Confusing versions.
Difficult maintenance.
No clear ownership.

Best practice:
Use Workspaces and Power BI Apps to centrally manage and distribute official reports.

3. Data Misuse and Excessive Permissions

Giving users unnecessary access can allow them to modify or distribute information that they should only view.

Example:
A normal business user receives editing permissions and accidentally changes an important report.

Best practice:
Use appropriate workspace roles and follow the least-privilege principle.

Overall Best Practice

A secure Power BI environment should generally use:

Workspace Roles → Control who can manage content

Power BI Apps → Control how approved content is distributed

RLS → Control which data each user can see

Together, these features improve security, governance, scalability, and data protection.
